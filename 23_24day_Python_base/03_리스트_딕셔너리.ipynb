```python
"""
<여러개의 변수값을 저장할 수 있는 변수(컬렉션 객체)>
 - 리스트(list), 딕셔너리(dict), 튜플(tuple), 기타(집합(set)...)
 - 꼭 알고 있어야하는 변수는 list와 dict


<리스트(list)>
 - 여러개의 값(타입)을 저장해 놓을 수 있는 변수를 의미함 : 리스트 변수라고도 칭함
 - 타입은 list 타입을 사용
 - **list를 의미하는 기호는 대괄호([])를 사용함(출력 결과에 대괄호 있으면 리스트)
 - 리스트 내에 저장된 여러개의 값들은 번호 및 순서를 가지고 있음
   (이 때 번호를 "인덱스(index) 번호"라고 칭함)
 - 인덱스 번호는 리스트가 관리하는 값들의 메모리 순서를 의미함
 - 리스트 내에 저장할 수 있는 값들은 : 일반변수 및 list, dict, tuple 등등 모두
 - **여러개의 값들을 처리하는 부분부터는 "메모리 관점"에서 이해해야 합니다.


<딕셔너리(dict)>
 - 여러개의 값을 가지고 다닐 수 있는 타입 중의 하나
 - 1개의 데이터(값)는 key와 value라는 값을 쌍으로 가지고 있음
 - 1개 데이터(값)에서 실제 값은 value에 저장되며,
   -> value에 저장되어 있는 실제 값을 추출하기 위해서는 key로 접근하여 추출합니다.
 - 딕셔너리(dict)를 표현하는 기호는 중괄호({})를 사용함
 - 리스트는 인덱스 번호를 이용하여 값(value)에 접근하고,
   딕셔너리는 key값을 이용해서 값(value)에 접근함
   따라서, 딕셔너리는 인덱스 번호에 의미를 두지 않음(순서 무관)
 - 딕셔너리에서 key값은 고유한 값만 사용해야함(중복 허용하지 않음)


<딕셔너리가 사용하는 key와 value에 저장할 수 있는 타입>
 - key : 정수, 실수 ,문자열 (주로 정수 또는 문자열을 사용)
 - value : 모든 타입 사용가능(정수, 실수, 문자열, bool, 리스트, 딕셔너리, 튜플....등등)
"""
"""
<튜플(tuple)>
 - 리스트(list)와 동일한 메모리 구조를 가지고 있음
 - 따라서, 리스트 사용법과 동일하게 사용 가능(인덱스 번호를 가지고 있음)
 - 리스트와의 차이점
   -> 소괄호() 기호를 사용함
   -> 수정을 할 수 없음 (수정을 하기 위해서는 삭제 후 다시 생성해야 함)
 - 튜플은 주로 정의해서 사용하기 보다는
   -> 임의 처리 결과를 받아서 사용하는 경우가 많음
   -> 예를 들어 zip() 함수의 결과 : [(값1, 값2), (값3, 값4)..]
   ->          여러 변수의 값 출력 결과 : a, b -> (a값, b값)
"""

```

```python
"""
- 이름 : name = []
- 나이 : age  = []
- 거주 지역 : area = []


1. 이름, 나이, 거주 지역을 각각 입력 받아서  -> 위의 각각의 리스트 변수에 저장
2. 특정 이름의 나이 데이터 값을 수정
3. 특정 이름의 모든 정보(이름, 나이, 거주지역) 삭제
4. 저장된 정보 모두 조회 : 모든 "이름, 나이, 거주지역" 출력
5. 특정 이름에 대한 한건 정보 조회 : "이름, 나이, 거주지역" 출력


1~5번은 무한반복을 통해 -> 입력 받고, 처리하고, 출력하기


-> 자유롭게 생각하여 프로그램 작성하시면 됩니다.
"""
```

```python
names = ["홍길동"]
ages  = [33]
areas = ["부산"]

while True:

    num_button = int(input("""숫자를 입력해주세요.
                            0. 종료 
                            1. 이름, 나이, 거주 지역을 각각 입력 받아서  -> 위의 각각의 리스트 변수에 저장
                            2. 특정 이름의 나이 데이터 값을 수정
                            3. 특정 이름의 모든 정보(이름, 나이, 거주지역) 삭제
                            4. 저장된 정보 모두 조회 : 모든 "이름, 나이, 거주지역" 출력
                            5. 특정 이름에 대한 한건 정보 조회 : "이름, 나이, 거주지역" 출력"""))

    # 0. 종료
    if num_button == 0:
        break

    # 1. 이름, 나이, 거주 지역을 각각 입력 받아서  -> 위의 각각의 리스트 변수에 저장
    if num_button == 1:
        name_input = input("이름을 입력하세요. 입력을 원치 않을 시 '종료'입력")
        age_input = int(input("나이을 입력하세요"))
        area_input = input("거주지역을 입력하세요")

        names.append(name_input)
        ages.append(age_input)
        areas.append(area_input)
        continue
     
    # 2. 특정 이름의 나이 데이터 값을 수정
    if num_button == 2:
        name_idx = names.index(input("나이를 바꿀 이름을 입력"))
        change_age = int(input("바꿀 나이을 입력하세요"))
        ages[name_idx] = change_age
        continue

    # 3. 특정 이름의 모든 정보(이름, 나이, 거주지역) 삭제
    if num_button == 3:
        name_idx = names.index(input("전체 삭제할 대상(이름) 입력"))
        del names[name_idx]
        del ages[name_idx]
        del areas[name_idx]
        continue

    # 4. 저장된 정보 모두 조회 : 모든 "이름, 나이, 거주지역" 출력
    if num_button == 4:
        for nm, ag, ar in list(zip(names, ages, areas)):
            print(f"이름 : {nm}, 나이 : {ag}, 거주지역 : {ar}")0
        continue

    # 5. 특정 이름에 대한 한건 정보 조회 : "이름, 나이, 거주지역" 출력
    if num_button == 5:
        name_idx = names.index(input("조회할 특정 이름을 입력"))
        print(f"이름: {names[name_idx]}, 나이: {ages[name_idx]}, 거주지역 : {areas[name_idx]}")
```

```python
### 딕셔너리 정의하기
# - 깨끗한 딕셔너리(데이터 없는 상태)
dict_1 = {}
print(dict_1)

# 주소값은 6개가 사용됨.
# 큰 메모리: dict_1, "list_data"의 value값은 리스트라 따로 생성
# 작은 메모리: -> dict_1 주소값에 해당하는 각 key값들 주소값
dict_1 = {"name":"홍길동", "age":23, "area":"부선 수성구", "list_data" : [1, 2, 3, 4, 5]}
print(dict_1)
```

```python
### dict_3 깨끗한 딕셔너리 변수 생성하기
# - 아래와 같이 딕셔너리 데이터 추가해주세요..
# - {1:"홍길동1", 2:"홍길동2", ...., 100:"홍길동100"}

dict_3 = {}
for i in range(0, 100, 1):
    i += 1
    dict_3[int(i)] = f"홍길동{i}"

print(dict_3)
```

```python
"""
* dict_3 딕셔너리 데이터를 이용해서..
    새로운 딕셔너리 생성하기 : 변수명 dict_4
  - dict_3의 key가 짝수이면 -> dict_4에 {"짝수" : [짝수 key에 대한 value값들만 리스트에 저장]}
  - dict_3의 key가 홀수이면 -> dict_4에 {"홀수" : [홀수 key에 대한 value값들만 리스트에 저장]}

  - dict_4 확인
"""
dict_4 = {}
# even_list = []
# odd_list = []
dict_4 = {"짝수" : [], "홀수" : []}

for k, v in dict_3.items():
    ### key 짝수
    if k % 2 == 0:
        # even_list.append(v)
        dict_4["짝수"].append(v)
        
    ### key 홀수
    else:
        # odd_list.append(v)
        dict_4["홀수"].append(v)
        

# dict_4["짝수"] = even_list
# dict_4["홀수"] = odd_list
print(dict_4)
```
