## 함수

```python
"""
<함수(C : function, JAVA : method, User Define Function)>
 - 자주 반복해서 사용되는 프로그램을
   -> 하나의 이름으로 정의해 놓은 후
   -> 호출해서 사용하는 방식("기능" 정의하고도 합니다.)


 - 미리 정의되어 라이브러리에서 제공되는 함수 : 내장함수라고 칭함
   -> 내장 함수의 이름은 -> 일반변수 이름으로 사용하면 안됩니다.
   
"""
```

```python
"""
### 더하기 연산을 수행하는 함수 생성하기
# <함수 작성 순서(정해진 규칙은 없음)>
#  1. 프로그램을 먼저 테스트 작성
#  2. 자주 반복해서 사용되는 부분을 찾기
#  3. 오류가 없는 상태로 테스트 수행
#  4. 자주 반복해서 사용되는 부분을 하나의 이름(함수명)으로 정의해서 관리하기
#  5. 자주 반복해서 "사용되던 부분"을 함수로 대체하기(함수 호출이라고 칭함)
#  6. 프로그램 테스트 하기 : 정의된 함수를 호출하여 실제 함수 실행
#                        : 이때 함수 내에서 처리가 잘 되는지 확인 가능

<함수 이름 작명>
- 함수 이름을 작명하는 규칙은 없으나, 현업에서 주로 사용하는 패턴은
   -> getter, setter, init_, view_ 등의 패턴을 주로 사용ㅎ암
- getter : 함수로부터 값을 리턴 받아야 하는 경우 getId, getPw와 같이 get~ 으로 시작
- setter : 함수를 이용하여 임의 값을 저장해야 하는 경우 setId, setPw와 같이 set~으로 시작
- init_  : 함수를 이용하여 임의 초기값 정의 또는 임의 함수를 최초에 시작하는경우
            init_function 과 같이 작명

- view_  : 시각화 또는 출력 등과 같이 함수를 이용하여 보여주는 기능을 수행하고자 할 때..

<함수의 형태>
- 매개변수 : 함수이름 뒤에 괄호에 들어가는 임의 변수 의미
           : 어떤값이 들어올지 모르지만, 호출하는 쪽에서 값을 넘겨줄 때 받아서 저장할 변수 의미

- 리턴(return) : 함수 처리 완료 후 처리 결과를 -> 호출한 쪽으로 전달하고자 할 때 사용



 1. 매개변수도 없고, 리턴값도 없는 형태
    def func():
        프로그램 처리
    -> 호출 방법 : func()

 2. 매개변수는 있고, 리턴값은 없는 형태
     def func(p_id, p_pw...):
        프로그램 처리
    -> 호출 방법 : func(asd1, dfw1)

 3. 매개변수는 없고, 리턴값은 있는 형태
     def func():
        프로그램 처리
        return 값
    -> 호출 방법 : rs_값1 = func()

 4. 매개변수도 있고, 리턴값도 있는 형태
      def func(p_id, p_pw...):
        프로그램 처리
        return 값
      -> 호출 방법 : rs_값1 = func(asd1, dfw1)

5. 매개변수 여러개를 리스트로 받는 형태
    def func(*data):
        프로그램처리
        return 있으면 추가
    -> 호출 방법 : func(1, 2, 3, 4, 5)
      -> *data = (1, 2, 3, 4, 5) 튜플 타입(리스트 형태)으로 받아서 처리

6. 매개변수 여러개를 딕셔너리 형태로 받는 형태
    def func(**data):
        프로그램처리
        return 있으면 추가
    -> 호출 방법 : func(name="홍길동", age=13, area="부산 남구")
      -> *data = {name="홍길동", age=13, area="부산 남구"} 딕셔너리 타입으로 받아서 처리

7. 함수에 default(기본값, 디폴트값, 생략가능 변수) 사용
     def func(p_id="1", p_pw="2") :
    -> 호출 : func(p_id="a001") -> p_pw는 기본값(디폴트값 사용됨)
"""
```

```python
### 덧셈 함수...
# - 임의값 2개를 함수에게 전달하면,
# - 함수는 2개의 값을 받아서,
# - 더하기 연산을 수행한 후
# - 결과값을 반환(리턴)...

def getSum(a, b):
    return a + b

total = 0
def getSum2(a, b):
    total = a + b
    return total

### 호출 : 테스트
print(getSum(10, 11))
print(getSum2(10, 15))
```

```python
"""
 - 2개 이상의 값을 전달 받아서,
 - 덧셈, 뺄셈, 곱셈 중에 한가지를 처리하는 연산 함수 정의
 - 연산된 결과 값은 반환합니다.
 - 반환받은 값을 변수에 저장 후 출력하기.
 - 함수 이름은 자유롭게
"""

def getSum(*data):
    input1 = input("덧셈, 뺄셈, 곱셈 중에 한 가지를 입력 하세요")

    if input1 == "덧셈":
        print(data[0])
        plus_total = data[0]
        for i in range(1, len(data), 1):
            plus_total += data[i]
        return(plus_total)

    if input1 == "뺼셈":
        print(data[0])
        minus_total = data[0]
        for i in range(1, len(data), 1):
            minus_total -= data[i]
        return(minus_total)

    if input1 == "곱셈":
        print(data[0])
        multiple_total = data[0]
        for i in range(1, len(data), 1):
            multiple_total *= data[i]
        return(multiple_total)
```

```python
### 구구단 출력 함수 만들기
"""
구구단에서
 - 시작단과 끝단은 임의 값으로 받아서 처리 (시작 매개변수, 끝값 매개변수) -> 중간은? for문 써서 채우기?
 - 곱할 숫자의 시작값과 끝값도 받아서 처리 (시작 매개변수, 끝값 매개변수) -> 중간은? for문 써서 채우기?
 <예시>
  3단부터 5단까지 1부터 7까지만 곱하고자 한다면
  -------[3단]-------
  3 x 1 = 3
   ...
  3 x 7 = 21
  -------[4단]-------
  4 x 1 = 4
    ...
  4 x 7 = 28
  -------[5단]-------
  5 x 1 = 5
    ...
  5 x 7 = 35
"""

# 시작단과 끝단은 임의 값으로 받아서 처리 (시작 매개변수, 끝값 매개변수) -> 중간은? for문 써서 채우기?
def getGugu():
  try:
    start = int(input("시작단의 초기값 입력하세요"))
    end = int(input("시작단의 끝값 입력하세요"))
    if start > end:
       print("초기값은 끝값보다 작아야 합니다.")
       return "" #return을 만나면 함수는 종료된다고 보면 됨.
  except:
    print("정수를 입력해주세요")

#- 곱할 숫자의 시작값과 끝값도 받아서 처리 (시작 매개변수, 끝값 매개변수) -> 중간은? for문 써서 채우기?

  try:
    start1 = int(input("곱할값의 초기값 입력하세요"))
    end1 = int(input("곱할값의 끝값 입력하세요"))
    if start1 > end1:
      print("초기값은 끝값보다 작아야 합니다.")
      return ""
  except:
    print("정수를 입력해주세요")

  for dan in range(start, end+1, 1) :
      print(f"============== [{dan} 단] ==============")
      for num in range(start1, end1+1, 1) :
          print(f"{dan} x {num} = {dan * num}")
  # return "" #위에서 return을 썼을 시에는 정상 출력인 경우에도 return을 넣어줘야 함.
```

## 02. 클래스(class)

```python
"""
<클래스(class)>
 - 객체(object)라고 칭함(객체지향 프로그램에서 사용되는 개념임)
 
 - 지금까지 사용한 일반변수, 여러값 가지고 다니는 변수, 함수 등을
   모아서 관리하는 메모리 공간(주소를 가짐)
   
 - 특정한 관리를 하나의 메모리 공간에 모아서 사용할 때 사용
   -> 예를 들어 회원관리라고 한다면, 회원 한명 한명에 대한 정보를 관리하는 클래스 생성
      --> 회원관리 클래스 내에 여러 기능들과 데이터를 관리하는 변수들을 정의하고,
      --> 외부에서 해당 클래스를 생성(메모리 주소 생성)하여 필요한 기능과 변수들을 이용함


<클래스 정의 방법>
 ### 클래스 이름의 첫글자는 대문자로 시작해야 합니다.(규칙)
 #   (함수는 주로 소문자로 시작 합니다.)
 class 클래스이름(MemberInfo) :
     변수정의


     ### 함수의 매개변수에 무조건 self를 넣어야 합니다.(규칙)
     # - 이때 self는 맴버를 의미합니다.(해당 클래스에 속해있다라는 의미)
     # - self가 붙은 함수를 맴버 함수라고 칭함(외부에서 접근 가능한 함수를 의미함)


     ### 생성자 함수는 생략가능
     #  - 단, self를 제외한 받아올 매개변수가 없는 경우 및
     #        처리할 문장이 없는 경우에
     def 생성자함수(self, 매개변수, ..) :
         처리 문장
         
     def 일반함수(self, 매개변수, ...) :
         처리 문장
"""

```

```python
### 사용자 정보 관리 클래스 정의하기 (껍데기 입니다.)
# - 생성하지 않으면 사용 못 합니다.
# - *** 클래스 이름의 첫글자는 "대문자"로 시작합니다.

class MemberInfo :
    ### *** 클래스 내에서 함수를 정의할 때는 무조건
    #       - 매개변수 첫번째에는 self를 붙여야 함.
    #         (받아올 매개변수가 없더라고 self는 붙여야 합니다.)
    #       - 외부에서 함수 호출시에는 self는 매개변수로 취급하지 않습니다.
    #         (없다고 생각하면 됩니다.)
    #       - self의 용도 : 클래스 내부 및 외부에서 어디서든지 접근 가능하도록 하기위한 속성입니다.

    ### 클래스 생성시 자동으로 호출되는 생성 함수
    def __init__(self):
        print("MemberInfo 클래스가 생성되었습니다.")

    ### 테스트 함수
    def viewPrint(self):
        print("viewPrint() 함수 호출")
        print(f"self.param = {self.param}")

    ### self 테스트
    def setParam(self) :
        print("setParam() 함수 호출")
        self.param = "홍길동"

#################################################################

    def setList(self, p_name, p_age, p_area):
        name.append(p_name)
        age.append(p_age)
        area.append(p_area)

    def viewAllList(self):
    #    name_age_area = zip(name, age, area)
        for nm, ag, ar in zip(name, age, area):
            print(f"이름은[{nm}]이고, 나이는 [{ag}]이며, 지역은 [{ar}]입니다. ")

    ### 임의의 이름을 이용하여 삭제하는 함수 정의하기
    # - 함수이름 : setDelList
    def setDelList(self, del_name):

        idx = name.index(del_name)

        del name[idx]
        del age[idx]
        del area[idx]
```

```python
### 클래스 생성하기
memInfo = MemberInfo()

### setParam() 함수 호출
memInfo.setParam()

### 점(.)의 의미 : 클래스 메모리로 접근하겠다는 의미("접근지정자 칭함")
memInfo.viewPrint()

### param 변수에 접근하기
print(memInfo.param)

### param 변수의 값 변경하기
memInfo.param = "이순신"
memInfo.viewPrint()
```

```python
### 클래스 생성하기
memInfo2 = MemberInfo()

### 실제 프로그램 시작 영역
name = []
age = []
area = []

### 저장함수 호출하기

memInfo2.setList("홍길동", 12, "부산")
setList("홍길송", 12, "부산")

### 출력함수 호출하기
memInfo2.viewAllList()

# -----------------------------
### setParam() 함수 호출
memInfo2.setParam()

print(memInfo.param)
print(memInfo2.param)
```
