## 01. 조건문

```python
"""
<조건문>
 - 2개의 값을 비교하고자 할 때 사용
 - 보통 조건 연산자(비교 연산자)를 이용하여 사용
 - 2개 이상의 조건을 만들 수 있음
 - 조건(비교) 연산자의 결과는 bool 타입(True 또는 False)의 결과 값을 반환 
                    : 크다, 작다, 크거나 같다, 작거나 같다, 같다, 같지않다
                    :  >     <        >=         <=       ==      !=
 - 규칙 : 조건문은 if로 시작해야 함
        : 조건 뒤에는 세미콜론(:) 넣어야 합니다. (처리 프로그램의 블록을 의미함)
        : 세미콜론(:) 밑에는 항상 들여쓰기로 처리 프로그램을 작성해야함
 - 작성방법 : if 값1 > 값2 :
                처리 프로그램
              elif 값1 > 값3 :
                처리 프로그램
              elif ... :
                 ....
              else :
                 위 조건이 모두 만족하지 않으면 처리
 - if ~ elif ~ else : if만 작성해도되며, if ~ else만 작성해도 됨
"""
```

```python
"""
아이디 "hong", 패스워드 "asdf"를 가지는 회원이 있습니다.
- 아이디와 패스워드를 입력 받았다고 가정하고,

- 입력한 아이디가 "hong"이고, 패스워드가 "asdf"이면
  -> "hong님 로그인 되었습니다"를 출력

- 아이디 또는 패스워드가 일치하지 않으면,
  -> "아이디 또는 패스워드가 일치하지 않습니다."를 출력
"""
```

```python
member1 = {"id" : "hong", "pw" : "asdf"}

id_input = input("아이디를 입력하세요")
pw_input =  input("패스워드를 입력하세요")

if (id_input == member1["id"]) and (pw_input == member1["pw"]):
    print(f"{member1['id']}님 로그인 되었습니다")

else:
    print("아이디 또는 패스워드가 일치하지 않습니다.")
```

```python
"""
- 나이를 입력 받아서
- 나이가 20세 미만이면 "청소년 할인"을 출력
         20세 이상이면서, 60세 미만이면 "일반인"을 출력
         60세 이상이면 "실버 할인"을 출력
         위 조건 모두 만족하지 않으면 "알수 없음"을 출력
"""
```

```python
age_input = int(input("나이를 입력해주세요"))

if age_input < 20:
    print("청소년 할인")

elif (age_input >= 20) and (age_input < 60):
    print("일반인")

elif (age_input >= 60) and (age_input < 110):
    print("실버 할인")

else:
    print("장난치지마세요")
```

```python
"""
과목 1개에 대한 점수를 입력 받아서...
 - 입력 받은 점수에 따른 학점을 출력하고자 합니다.
 * 100~95점은 "A+" 출력
   94~90점은  "A0" 출력
   89~85점은 "B+" 출력
   84~80점은 "B0" 출력
   79점이하 점수는 "F" 출력
"""
```

```python
score = int(input("점수를 입력:"))

if (score >= 95) and (score <= 100):
   print("A+")
elif (score >= 90) and (score < 95):
   print("A0")
elif (score >= 95) and (score < 90):
   print("B+") 
elif (score >= 80) and (score < 85):
   print("B0")
else:
   print("F")
```

## 02. For 문

```python
### for문 기본 사용법

# - 튜플(touple)
for num in (1, 2, 3):
    print(num)
t = (1, 2, 3)

for i in range(0, len(t), 1):
    print(t[i])

# - 리스트(list)
l = [4, 5, 6]
for num in l:
    print(num)

for i in range(0, len(l), 1):
    print(l[i])

# - 집합(set)
s = {7, 8, 9}
for num in s:
    print(num)

### 집합은 인덱스 번호로 접근 안됨
# for i in range(0, len(s), 1):
#   print(s[i])
```

```python
# - zip() 함수 이용 : 여러개의 값을 저장하는 타입(리스트, 튜플, 집합)에 대해서
#                     각 인덱스 위치의 값끼리 튜플로 묶어서 전체 리스트로 처리됨
aa = (1, 2, 3)
bb = [4, 5, 6]
cc = [7, 8, 9]
# cc = {7, 8, 9} : 집합 함수는 인덱스가 없어서 랜덤하게 출력됨. zip하면 랜덤하게 묶임

zip_abc= list(zip(aa, bb, cc))
print(zip_abc)

aaa, bbb, ccc = (1, 2, 3)
print("-"*30, aaa, bbb, ccc)

for a, b, c in zip_abc:
    print(a, b, c)
```

```python
"""
1부터 100까지의 총합을 출력하기
- 마지막 한번만 print()함수로 총합만 출력
"""
i_total = 0
for i in range(1, 101, 1):
    i_total += i
    
print(i_total)
```

```python
"""
1부터 100까지의 정수값에서
- 짝수의 종합을 마지막 한번만 print()함수로 출력
"""
i_total = 0
for i in range(1, 101, 1):
    if i % 2 == 0:
        i_total += i
    
print(i_total)
```

```python
### 2024년 1월 1일부터 2025년 12월 31일까지의 값을 출력
# - 월은 1~12월까지
# - 모든 월에 대한 일자는 1~31일까지 있는 것으로 가정합니다.
# 출력예시 :
#  "2024-01-01"
#  "2024-01-02"
#    ....
#  "2025-12-31"

for yy in ["2024", "2025"]:
    for mm in range(1, 13, 1):
        for dd in range(1, 32, 1):
            if (mm < 10) and (dd < 10):
                print(f"{yy}-0{mm}-0{dd}")
            elif (mm < 10):
                print(f"{yy}-0{mm}-{dd}")
            elif (dd < 10):
                print(f"{yy}-{mm}-0{dd}")
            else:
                print(f"{yy}-{mm}-{dd}")
```

```python
for yy in ["2024", "2025"]:
    for mm in range(1, 13, 1):
        for dd in range(1, 32, 1):
            ### 포맷팅할 때 실수의 소수점은 f로, 정수는 d로 표현
            # - 02d : 두자릿수가 아닌 경우 앞에 0을 채워라
            # - d 앞 숫자는 자릿수(메모리공간)를 의미
            print(f"{yy}-{mm:02d}-{dd:02d}") 
            
            # - zfill(2) : 2자리 공간을 만들고, 오른쪽부터 채우면서, 빈공간은 0(zero)으로 채우라는 함수
            print(f"{yy}-{str(mm).zfill(2)}-{str(dd).zfill(2)}") 
            
```

## 03. While 문

```python

"""
<while문>
 - 조건이 있는 경우와 없는 경우로 나누어서 반복을 수행 할 수 있음
 - for문의 경우에는 숫자의 범위 패턴을 이용해서 반복했다면,
   -> while문은 숫자의 범위 보다는, "종료조건"을 만족할 때까지 반복을 수행합니다.
 - while문은 가급적 종료하는 방법을 작성해야 합니다.
   -> "종료는 조건과 break 명령"을 통해서 주로 처리됩니다.
 - 종료하는 방법을 작성하지 않으면 -> 무한반복(무한루프) 됩니다.
 - for문으로 작성한 것들은, 일반적으로 while문으로도 작성 가능합니다.


 - 종료조건 문법
    while 종료조건 :
      문장(프로그램 처리)


 - 무한루프 문법
    while True :
      if 종료조건 :
        break
       
      문장(프로그램 처리)
"""

```

```python
"""
 - 종료조건 문법
    while 종료조건 :
      문장(프로그램 처리)

"""
a = 0
b = 5
while a <= 5 :
    print(f"a = {a}")
    a += 1

```

```python
a = 0
b = 5
while True :
    ### 종료 조건
    if a > b :
        print("강제 종료")
        break

    print(f"a = {a}")
    a += 1
```

```python
"""
while문을 이용하여 1부터 100까지의 총합을 출력해주세요.
- 출력을 위한 print()함수는 마지막에 한번만 사용
"""

total = 0
i = 0
while i < 100:
    i += 1
    total += i
print(total)
```

```python
i = 0
total = 0
while True:
    i += 1
    total += i
    if i == 100:
        break
print(total)
```

```python
num1 = 2
num2 = 0

while True:
    num2 += 1
    if num2 > 9:
        num1 += 1
        num2 = 0
        continue
    if num1 > 9:
        break
    print(f"{num1} x {num2} = {num1*num2}")

print(f"[{'+'*150}]")

num1 = 2
num2 = 0
while num1 <= 9:
    num2 += 1
    if num2 > 9:
        num1 += 1
        num2 = 0
        continue
    print(f"{num1} x {num2} = {num1*num2}")
```
