### LLM 모델을 활용한 랭체인 테스트

```python

### Ollama 설치 확인 및 초경량 모델 구동 테스트
# - Ollama가 잘 작동하는지 확인하기 위해, 현재 공개된 모델 중
#   가장 용량이 작고 가벼운 축에 속하는 qwen2.5:0.5b (약 400MB)를 활용하여 테스트


### <단계 1: 터미널에서 모델 다운로드 및 실행 확인>
#  - 0.5B 사이즈의 초경량 Qwen 모델 다운로드 및 실행
#  - 터미널에서 아래 명령 실행하여 모델 다운로드 하기
#    >ollama run qwen2.5:0.5b


#  - 처음 실행하면 모델을 다운로드(pull)하느라 시간이 조금 걸림
#  - CLI에서 정상 작동을 확인
#   -> 다운로드가 완료되면 ">>>" 프롬프트가 보임
#   -> "안녕"이라고 입력하여 -> [답변]이 나오는지 확인한 후,
#   -> "/bye"를 입력하여 CLI 프롬프트 종료하기


### <단계 2: LangChain + Ollama 연동 확인>
#  - 랭체인(LangChain)을 통해 Ollama 모델을 호출할 수 있는지 테스트
```

```python
### 라이브러리 불러들이기
# - 올라마 구동되고 있어야 정상 실행됨
from langchain_ollama import ChatOllama
```

```python
print("Ollama 모델에 연결을 시작합니다...")

### 예외 처리(외부에서 발생하는 문제를 대처하기 위해서)
# try : 정상 작동 중에 예외가 발생하면, except에서 처리
try:
    # 1. LLM 객체(클래스) 생성(다운로드 받은 모델 이름 지정)
    llm=ChatOllama(
        # - 모델 지정하기
        model="qwen2.5:0.5b",

        # - 답변의 창의성 조절(0에 가까울수록 일관된 답변 작성)
        temperature=0.1
    )
    #print(llm)

    # 2. 모델에 질문 전달하기 위해 질문 정의하기
    prompt = "안녕? 너는 누구니? 아주 짧게 한 문장으로 답변해줘."
    print(f"[질문] : {prompt}")

    # 3. 모델에게 질문을 전달하고 답변 받아오기 위한 처리
    #  - invoke : 직렬화(나의 프로그램 pc에서 올라마 서버로 연결하여 처리-데이터 자동 전달/받기)
    response = llm.invoke(prompt)
    # print(f"response : {response}")

    # 4. 응답받은 내용(답변) 출력하기
    # - 응답 결과 내에서 content 변수에 응답(답변)이 저장되어 있음
    print(f"[답변-사용모델({llm.model})] : {response.content}")


except Exception as e:
    ### except 외부에서 발생하는 오류에 대한 처리. 문법적으론 오류가 없음.
    print(f"\n오류가 발생하였습니다. : {e}")
    print("Ollama 서버가 잘 작동중인지 확인")
    print("사용하는 모델을 다운로드하였는지 확인")
```

```python
"""
<랭체인(LangChain) 이란>
    - 랭체인 정의
        : "언어 모델(Language Model)을 다른 유용한 도구(or 기능)들과 연결(Chain)하여
           실용적인 or 자동화된 애플리케이션을 만드는 기술(프레임워크)" 이라고 정의


    - Lang의 의미 : Language Model(언어 모델)의 줄임말
        -> 똑똑하지만 혼자서는 아무것도 할 수 없는 AI 두뇌?
   
    - Chain의 의미 : 사슬, 연결고리 (보통 파이프라인-Pipeline-이라고도 불리는 개념)
        -> 외부 데이터나 도구, 여러 작업 순서 등을 => AI 모델과 매끄럽게 연결하는 것
        -> 즉, 모델이 움질일 수 있도록 => 자동화된 처리 흐름을 만들어 준다는 개념


    (예시)
    - 각기 다른 기능(프롬프트 정의, 모델 생성, 출력 등)들을
        -> 마치 레고 블록을 조립(|)하듯 매끄럽게 연결하여
        -> 하나의 완성된 처리 흐름(Chain == Pipeline)을 만드는 것


<랭체인을 사용하는 3가지 핵심 목적>
    1. 외부 데이터와의 연결(RAG, Retrieval-Augmented Generation, 검색 기반 생성)
        - AI는 자신이 학습한 과거의 데이터만 알고 있으며,
            -> 모르는 것도 아는 것처럼 꾸며내는 '환각(Hallucination)' 현상을 보임


        - 랭체인을 사용하면 AI가 답변을 하기 전에
            -> 외부의 "특정 문서"나 "데이터베이스"를 먼저 읽어보도록 지시할 수 있음
       
        - (예시)
            -> 방대한 국가 법령이나 복잡한 판례 같은 특정 데이터를 AI에게 인식 시켜주고
            -> "이 문서 안에서만 정답을 찾아서 대답해"라고 명령하는 고도화된 챗봇(RAG)을 만들 때
            -> 랭체인이 핵심적인 연결 역할을 수행함


    2. 여러 작업의 자동 연결 (Chaining)
        - 사람이 일을 할 때는 여러 단계를 거치게됨
       
        - "자료를 검색한다 ➔ 핵심을 요약한다 ➔ 영어로 번역한다 ➔ 이메일로 보낸다"처럼
            -> 랭체인은 이름(Chain) 그대로, AI가 수행해야 할 여러 단계의 명령(기능)들을
            -> 하나의 사슬처럼 매끄럽게 연결해 주는 역할을 함


        - 랭체인 공식 용어로는
            -> LCEL(LangChain Expression Language, 랭체인 표현 언어)라고 칭함


            <LCEL 용어의 의미적 해석>
                - LangChain (랭체인)
                    : 랭체인 프레임워크 안에서


                - Expression (표현/표현식)
                    : 복잡한 코드를 길게 나열하지 않고,
                      -> 기호(|)를 사용해 간단한 수식처럼 '표현'하는
                     
                - Language (언어/문법)
                    : 랭체인만의 독자적인 작성 규칙(문법)


    3. 기억력 부여 (Memory)
        - 기본적인 AI 모델은 사용자와 나눈 방금 전의 대화도 쉽게 잊어버림
        - 랭체인은 대화의 맥락을 저장하고 관리하는 '기억력' 모듈을 제공하여,
            -> AI가 사용자의 이전 질문과 상황을 기억하고 자연스럽게 대화를 이어갈 수 있음
"""
```

```python
### 프롬프트 : system을 의미함
# - 프롬프트에게 사전에 역할을 부여
# - 질문에 대한 목적에 맞는 응답 규칙 정의


sys_cmd = """
          주어진 무장의 감정을 긍정, 부정, 중립으로만 분류합니다.
          분류에 대한 추가 설명을 하지 않습니다.


          그리고 아래 <긍정 단어>와 <부정 단어>에 포함된 단어들이
          사용자가 입력한 문장에 포함되어 있으며, 참고해서 분류 합니다.

          <긍정 단어>
          꿀잼, 짜릿

          <부정 단어>
          밉상, 노잼
"""
sys_cmd
```

### 2. 프롬프트 템플릿 정의하기

```python
### 문장 감정 분류를 위한 프롬프트 템플릿 정의하기
# - 보통 프롬프트 엔지니어링이라고 칭하는 부분입니다.
# - 챗봇의 프롬프트는 역할이 구분되어 있으며,
#   역할에 맞게 프롬프트를 정의합니다.
# - 역할 : 모델(system, 챗봇 자기 자신), 사용자(user, 질문자)라는 이름을 주로 부여하여 정의

prompt_template = ChatPromptTemplate.from_messages([
    ### system(챗봇 자기 자신)에게 사전에 정의한 규칙 명령하기(역할 부여)
    ("system", sys_cmd),

    ### user(사용자,질문자)가 입력한 프롬프트 문장({sentence})을 전달받는 역할 부여
    # - sentence : 미리 질문을 출력할 변수명을 정의해 놓습니다.
    ("user", "문장 : {sentence}\n 분류 : ")
])

prompt_template
```

#### 3. LLM 모델 정의하기

```python
### LLM 모델 정의
# - 모델 : ollama의 mistral 모델 사용
#   -> ollama 서버에서 모델을 다운로드하여 실행해 놓아야 함
#   -> ollama run mistral
#   -> mistral : ollama에서 제공하는 모델로,
#              : 빠른 응답 속도와 효율적인 성능을 제공하는 모델임


# - temperature : 모델이 답변을 생성할 때의 창의성 정도를 조절하는 매개변수로,
#               : 낮은 값은 더 결정적이고 일관된 답변을 생성하는 반면,
#               : 높은 값은 더 다양하고 창의적인 응답을 생성함.
#               : 일반적으로 0.1에서 1.0 사이의 값을 사용.  
#     ** 데이터 전처리나 메타데이터 추출처럼
#        - '정해진 양식에 맞춰 정확한 답변'을 받아야 하는 작업이라면,
#        - temperature=0.1 (또는 0)으로 아주 낮게 유지하는 것이
#           -> 모델의 환각 현상을 줄이는데 효과적일 수 있음


# - max_tokens : 모델이 생성할 수 있는 최대 토큰 수를 제한하는 매개변수로,
#               : 응답의 길이를 제어하는 데 사용됨
```

```python
### 사용할 모델 : mistral 모델
llm = ChatOllama(
    model="mistral",
    temperature=0.1,
    max_tokens=100
)
llm
```

### 4.모델이 응답할 내용을 요약하여, 출력을 처리

```python
### 모델이 응답한 내용을 문자열로 반환하는 클래스
# - 최종 출력 결과를 문자열로 변환(파싱-Parsing)
parser = StrOutputParser()
parser
```

#### 5. 프롬프트 > 모델 > 결과출력의 과정을 연결(Chain)하기

```python
### 파이프(Pipe) 연산자를 이용하여 체인(chain) 흐름 정의(랭체인 적용)
# - 파이프(|)로 연결된 기능들은 순서대로 역할을 수행하고,
# - 처리된 데이터 값들은 자동으로 다음 연결로 전달 및 받아서 처리됨
#   -> 내부적으로 데이터 전달 등의 모든 일괄 처리는 Ollama 프레임워크가 자동으로 처리해줌


chain = prompt_template | llm | parser
chain
```

### 6. LLM 챗봇을 통해 추론(예측) 수행하기

```python
### 추론(예측)에 사용할 질문 정의하기
# - 실제로는 사용자로부터 입력을 받아서 사용하게됩니다.
"""
    "영화가 재미 없었어요~",
    "이 음식은 맛있어요....",
    "서비스가 별로 였어요~~~",
    "음식은 괜찮았던 것 같아요.",
    "오늘 축구는 짜릿했어요~~~",
    "이 영화는 재미있는것 같기도 하고, 별로 인것 같기도해~",
    "영화 너무 꿀잼이에요~",
    "영화가 노잼이에요~",
    "이 영화 보지 마셈~~"

"""

request_sentences = "영화가 노잼이야."
request_sentences = "영화가 맛도리잼이야."
request_sentences = "영화가 마꿀마잼이야."
request_sentences
```

```python
### 챗봇 기능 사용하기 -> 체인(Chain) 사용
# - chain.invoke() : 체인에 데이터를 전달하여 실행하는 메서드로,
#   -> 체인에 정의된 순서대로 프로세스를 처리하여 최종 결과를 반환함
#   -> 랭체인(파이프라인으로 연결된 기능들)의 시작점을 의미함

### sentence : prompt_template의 user에 정의가 미리 되어 있음
# - sentences에 정의한 문장들을 하나씩 체인에 전달하여,
#   -> prompt_template에서 정의한 프롬프트 규칙에 따라 system에게 명령을 전달하고,
#   -> llm이 명령에 따라 text를 생성하여 응답하게 되고,
#   -> parser가 llm의 응답에서 text만 파싱하여 최종적으로 문자열로 반환하게 됨
# - 최종적으로 각 문장에 대한 감정 분류 결과가 출력됨

result = chain.invoke({"sentence" : request_sentences})
print(f"{result}")
```

```python
### 프롬프트 : system을 의미함
# - 프롬프트에게 사전에 역할을 부여
# - 질문에 대한 목적에 맞는 응답 규칙 정의


sys_cmd = """
          주어진 무장의 감정을 긍정, 부정, 중립으로만 분류합니다.
          답변은 "긍정", "부정", "중립"으로중 하나만 답변하도록 합니다.
          분류에 대한 추가 설명을 하지 않습니다.


          그리고 아래 <긍정 단어>와 <부정 단어>에 포함된 단어들이
          사용자가 입력한 문장에 포함되어 있으며, 참고해서 분류 합니다.

          <긍정 단어>
          꿀잼, 짜릿

          <부정 단어>
          밉상, 노잼
"""
sys_cmd
```

```python
### 문장 감정 분류를 위한 프롬프트 템플릿 정의하기
# - 보통 프롬프트 엔지니어링이라고 칭하는 부분입니다.
# - 챗봇의 프롬프트는 역할이 구분되어 있으며,
#   역할에 맞게 프롬프트를 정의합니다.
# - 역할 : 모델(system, 챗봇 자기 자신), 사용자(user, 질문자)라는 이름을 주로 부여하여 정의

prompt_template = ChatPromptTemplate.from_messages([
    ### system(챗봇 자기 자신)에게 사전에 정의한 규칙 명령하기(역할 부여)
    ("system", sys_cmd),

    ### user(사용자,질문자)가 입력한 프롬프트 문장({sentence})을 전달받는 역할 부여
    # - sentence : 미리 질문을 출력할 변수명을 정의해 놓습니다.
    ("user", "문장 : {sentence}\n 분류 : ")
])

prompt_template
```

```python
llm = ChatOllama(
    model="qwen2.5:0.5b",
    temperature=0.1,
    max_token=100
)
parser = StrOutputParser()
chain = prompt_template | llm | parser

print(llm.model)
```
