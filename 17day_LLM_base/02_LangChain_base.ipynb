### 라이브러리 정의

```python
### 프롬프트 템플릿
from langchain_core.prompts import ChatPromptTemplate

### 올라마 라이브러리
from langchain_ollama import ChatOllama

### 응답 출력 파서 라이브러리
from langchain_core.output_parsers import StrOutputParser
```

```python
"""
1. 챗봇(System)에게 역할 부여하기(기존 역할 그대로 정의)
2. 프롬프트 템플릿 정의
3. 모델 정의(미스트럴 사용)
4. 출력 파서 정의
5. 랭체인 정의(변수명 chain)
"""
```

```python
# 1. 프롬프트 정의
sys_cmd = """
          주어진 무장의 감정을 긍정, 부정, 중립으로만 분류합니다.
          분류에 대한 추가 설명을 하지 않습니다.


          그리고 아래 <긍정 단어>와 <부정 단어>에 포함된 단어들이
          사용자가 입력한 문장에 포함되어 있으며, 참고해서 분류 합니다.

          <긍정 단어>
          꿀잼, 짜릿

          <부정 단어>
          밉상, 노잼
"""

# 2. 프롬프트 템플릿 정의
prompt_template = ChatPromptTemplate.from_messages([
    ### system(챗봇 자기 자신)에게 사전에 정의한 규칙 명령하기(역할 부여)
    ("system", sys_cmd),

    ### user(사용자,질문자)가 입력한 프롬프트 문장({sentence})을 전달받는 역할 부여
    # - sentence : 미리 질문을 출력할 변수명을 정의해 놓습니다.
    ("user", "문장 : {sentence}\n 분류 : ")
])

# 3. 모델 정의
llm = ChatOllama(
    model="mistral",
    temperature=0.1,
    max_tokens=100
)

# 4. 출력 파서 정의
parser = StrOutputParser()

# 5. 랭체인 정의
chain = prompt_template | llm | parser
chain
```

### 사용자로부터 질문 받기(또는 질문할 목록 정의)

```python
request_sentences = [
    "영화가 재미 없었어요~",
    "이 음식은 맛있어요....",
    "서비스가 별로 였어요~~~",
    "음식은 괜찮았던 것 같아요.",
    "오늘 축구는 짜릿했어요~~~",
    "이 영화는 재미있는것 같기도 하고, 별로 인것 같기도해~",
    "영화 너무 꿀잼이에요~",
    "영화가 노잼이에요~",
    "이 영화 보지 마셈~~"
]
```

```python
# - invoke() : 단일 입력을 처리하여 -> 결과를 반환하는 함수

### 질문 하나만 처리하는 방식
for sentence in request_sentences:
    response = chain.invoke({"sentence" : sentence})

    print(f"[질문] : {sentence}")
    print(f"[답변] : {response}\n")
```

```python
# - batch()  : 여러 입력을 동시에 처리하여 -> 결과를 반환하는 함수

### 질문 여러개를 동시에 처리하여 결과를 반환하는 방식
# 여러 질문 정의
sentence1 = "영화가 재미 없었어요~"
sentence2 = "이 음식은 맛있어요...."
sentence3 = "서비스가 별로 였어요~~~"
sentence4 = "음식은 괜찮았던 것 같아요."
sentence5 = "오늘 축구는 짜릿했어요~~~"
sentence6 = "이 영화는 재미있는것 같기도 하고 별로 인것 같기도해~"
sentence7 = "영화 너무 꿀잼이에요~"
sentence8 = "영화가 노잼이에요~"
sentence9 = "이 영화 보지 마셈~~"

### 여러개의 질문 동시 처리(batch())
response_list = chain.batch([
    {"sentence" : sentence1},
    {"sentence" : sentence2},
    {"sentence" : sentence3},
    {"sentence" : sentence4},
    {"sentence" : sentence5},
    {"sentence" : sentence6},
    {"sentence" : sentence7},
    {"sentence" : sentence8},
    {"sentence" : sentence9}
])

### 결과값 출력하기
response_list
```

```python
# - stream() : 입력을 토큰 단위로 스트리밍하여 -> 생성된 토근을 실시간으로 반환하는 함수
            # (결과를 완전히 생성하여 한번에 출력하기 전에, 생성된 각 토근을 하나씩 실시간으로 받아 볼수 있음)   


### 챗봇(System)에게 답변의 길이를 길게 할 수 있도록 정의
sys_cmd = "주어진 문장의 주제에 대해 짧게 한국어로 설명을 해줘."

### 프롬프트 템플릿 새로 정의
prompt_template = ChatPromptTemplate.from_messages([
    ("system", sys_cmd),
    ("user", "주제:{topic}\n설명")
])

### 프롬프트 템플릿이 수정되면, 체인도 다시 실행해야함
chain = prompt_template | llm | parser

### stream() 기능 사용하기
# - 질문 정의
request = "대규모 언어 모델(LLM)"
for token in chain.stream({"topic" : request}):
    print(token, end="", flush=True)
```

```python
def llm_model(m):
    # 3.
    llm = ChatOllama(
        model=m,
        temperature=0.1,
        max_tokens=100
    )
    return llm
```

### 이전 응답의 결과(체인=Chain)를 연결하여 사용하기

```python
### 이전 응답의 체인(Chain)
# - 이전 질문의 응답을 전달 받아서 계속해서 다음 질문을 이어나가는 방식

##################[첫번째 프롬프트 체인 처리(영문으로 응답 받기)]##################
# 1. 챗봇(system) 사전 역할 정의(영문 응답 역할)
sys_cmd = "{topic}에 대해서 짧게 한줄로 영어로 대답해줘."

# 2. 프롬프트 템플릿 정의 : 변수명은 prompt
# - from_template() : 프롬프트 명령에 대해 즉시 실행(system 규칙없이 작성)
prompt = ChatPromptTemplate.from_template(sys_cmd)

# 3. 출력 파서 정의 : 변수명은 parser
parser = StrOutputParser()

# 4. 랭체인(LangChain) 연결 구성 : 변수명은 eng_chain
eng_chain = prompt | llm | parser


############[두번째 프롬프트 체인 처리(응답받은 영문을 한국어로 번역하기)]###########
# 1. 챗봇(system) 역할 정의(영문을 한국어로 번역)
sys_cmd = "이 대답을 한국어로 번역해줘. : {answer}"

# 2. 프롬프트 템플릿 정의 : from_template() 함수 사용
answer_prompt = ChatPromptTemplate.from_template(sys_cmd)

# 3. 출력 파서 : 이미 위에 정의한 변수 그대로 사용해도 됨

# 4. 랭체인(LangChain) 연결 구성
# - 첫번째 프롬프트(prompt)의 응답 결과를 "answer"에 담아서 
#   다음 프롬프트(answer_prompt)에 전달하여 응답 처리 진행
kor_chain = {"answer" : prompt} | answer_prompt | llm | parser
```

```python
### 실행 시키기.
# - 마지막 체인을 사용함. 마지막 체인에도 첫번째 프롬프트가 있기 때문
# - 딕셔너리 타입의 topic key에 담아서 kor_chain에게 넘김
# - kor_chain의 시작 체인은 prompt 기능이 시작되며,
# - prompt 기능에서 topic을 받아서 처리하게 변수 처리 되어있음.
result = kor_chain.invoke({"topic":"대규모 언어 모델(LLM)"})
# kor_chain이 시작점. {"answer" : prompt} 에서 parser값은 prompt에 바로 보냄. 출력한 것이 아님.
print(result)
```

### 여러개의 체인(Chain)을 [병렬]로 처리하여 각각 출력하기

```python
"""
<여러개의 체인 병렬 처리 클래스>
 - RunnableParallel : 여러개의 체인을 병렬로 실행하여 효율성을 높이는 방법 처리
 - 각 체인의 결과(응답)을 각각 출력 할 수 있음

 - 사용패키지 : langchain_core.runnables
 - 사용클래스 : RunnableParallel
"""
```

```python
### 라이브러리 불러들이기
from langchain_core.runnables import RunnableParallel
```

```python
### <체인1> 영문으로 토픽 설명
# - {topic} : invoke("topic" : "사용자질문")
eng_chain = (
    ChatPromptTemplate.from_template("{topic}에 대해 짧게 영문으로 설명해줘.")
    | llm
    | parser
)

### <체인2> 국문으로 토픽 설명
kor_chain = (
    ChatPromptTemplate.from_template("{topic}에 대해 짧게 한국어로 설명해줘.")
    | llm
    | parser
)

### 병렬 랭체인 연결 구성하기(병렬 처리 = 각각 동시에 처리)
# - RunnableParallel()의 응답 결과는 "딕셔너리 타입"으로 반환됨
# - -> {"eng" : "영문 응답결과", "kor":"한국어 응답결과"} 형태로 반환됨
# - RunnableParallel(eng=eng_chain, kor=kor_chain) : eng와 kor 속성은
#   -> 응답결과 딕셔너리 타입의 key로 사용됨(이름은 자유롭게 정의 가능)
parallel_chain = RunnableParallel(eng = eng_chain, kor = kor_chain)

### 질문을 이용하여 병렬처리 후 응답 받아오기
# - invoke({"key 프롬프트에서 받을 변수명" : "value 요청 질문이 들어가는 부분"})
"""
1. 사용자의 질문("대규모 언어 모델(LLM)")을 
    -> 딕셔너리 타입으로 {"topic" : "대규모 언어 모델(LLM)"} parallel_chain에게 전달함
2. parallel_chain이 작동 시작함
    -> 딕셔너리 타입의 {"topic" : "대규모 언어 모델(LLM)"} 데이터를 전달 받음
    -> 전달받은 딕셔너리 타입의 key와 value의 값을 각각 eng_chain과 kor_chain에게 전달
3. eng_chain과 kor_chain은 딕셔너리 타입의 key와 value의 값을 전달 받음
    -> eng_chain의 from_template("{topic}에 대해 짧게 영문으로 설명해줘.") 문자열 내부에 {topic}(딕셔너리 key)에
       전달 받은 value의 값을 삽입하게 됨

    -> kor_chain의 from_template("{topic}에 대해 짧게 영문으로 설명해줘.") 문자열 내부에 {topic}(딕셔너리 key)에
       전달 받은 value의 값을 삽입하게 됨
"""
result = parallel_chain.invoke({"topic":"오늘 날씨"})
result
```
