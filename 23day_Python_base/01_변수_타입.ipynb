## 01. 변수타입

```python
### 변수 이름 규칙(정해진 규칙 입니다. 모든 프로그램 언어 동일)
# - 숫자로 시작하지 않기
# - 띄어쓰지 않기
# - 라이브러리에서 사용되는 이름은 사용 못하는 경우가 많음
#   (파이썬 내장 함수 이름이나 내부적으로(라이브러리) 정의된 이름은 피하기)
#   (코드 내에서 파이썬 내장 함수가 작동하지 않게 됨)
# - 한글은 사용하지 않기
# - (변수명 + 문자열) 안됨 -> abc + "def"


### 일반적 이름 작명방법
# - 저장되는 값의 성격 또는 타입에 맞는 변수명을 사용
# - 의미를 부여해야 하는 경우에는 언더라인(_)을 이용해서 단어적 의미 조합으로 가능
# - 가장 앞에 언더라인(_)을 사용할 수 있음
# - 변수명이 대문자로 쓰이는 경우 : 변하지 않는 값을 정의할 때 주로 대문자 변수명 사용
```

```python
### (해석) : a라는 메모리 공간(주소값)에 777 값 저장
a = 777
```

```python
"""
- 정수 타입 : int, 타입변환(형변환) int()
- 실수 타입 : float, 타입변환(형변환) float()
- 문자열 타입 : str, 타입변환(형변환) str()
- 참거짓 타입 : bool
"""
```

```python
### 문자열 안에 숫자만 있는 경우 int 타입으로 변환 가능
p_int = "1234"
a = 45

print(a * p_int)

print(a + int(p_int))

print(str(a) + p_int)
```

### 산술 연산자

```python
"""
<산술 연산자>
 - * 더하기 : +
     : 문자열과 문자열 합치기 가능

 - * 빼기 : -

 - * 곱하기 : *
     : 문자열 같은 값 여러개 만들 때도 사용

 - * 나누기 : /
 - 나눈 결과의 몫 : //

 - * 나눈 결과의 나머지 : %

 - 제곱승 : **

 - +=, -=
   a += 1(a = a + 1)
   a -= 1(a = a - 1)
"""
```

### 문자열 표현 방식(format)

```python
name = "홍길동"
age = 10
rate = 0.90999999
addr = "부산 수영구 광안로" 
str_message = "이름은" + name + "나이는" + str(age) + "주소는" + addr
str_message = "이름은 {}, 나이는 {}, 주소는 {}".format(name, age, addr)
str_message = "이름은 {0}, 나이는 {1}, 주소는 {2}".format(name, age, addr) # format(내 값들은 index 별로 불러오기 가능)


print(str_message)
print(f"이름은 {name}, 나이는 {age}, 주소는 {addr}, 비율은 {rate: .3f}") # :.5f 5번째 자리에서 반올림 후 5번째까지 표시
print("이름은 %s, 나이는 %s, 주소는 %s" %(name, age, addr))
```
